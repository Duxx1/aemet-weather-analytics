# Exploratory Data Analysis - AEMET Climatology
## Badajoz Province · Monthly Records · 2000-2025

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)
plt.rcParams["figure.figsize"] = (14, 5)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
sns.set_palette("tab10")
```

```python
DATA_DIR = Path("../data/processed")

df = pd.read_parquet(DATA_DIR / "climatology_monthly.parquet")

print(f"Shape: {df.shape}")
print(f"\nData reange: {df['fecha'].min()} -> {df['fecha'].max()}")
print(f"Stations:   {df['indicativo'].nunique()}")
print(f"\nDtypes:\n{df.dtypes}")
```

```python
df.head(10)
```

## 2. Dataset Overview

```python
# Percentage of missing values per column, sorted descending
missing = ( df.isnull().sum() / len(df) * 100 ).sort_values(ascending=False)
# Keep only columns that have missing values
missing = missing[missing > 0].rename("missing_%")

fig, ax = plt.subplots(figsize=(8, 10))
missing.plot(kind="barh", ax=ax, color="steelblue")
ax.set_xlabel("Missing (%)")
ax.set_title("Missing values by column")
# Reference line: columns above this are too sparse for reliable analysis
ax.axvline(50, color="red", linestyle="--", linewidth=0.8, label="50%")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# First date, last date and total record count per station
coverage = (
    df.groupby("indicativo")["fecha"]
    .agg(["min", "max", "count"])
    .rename(columns={"min": "first", "max": "last", "count": "n_records"})
    .sort_values("n_records", ascending=False)
)

print(coverage.to_string()) # avoid truncating rows
```

```python
# Pivot table: rows = stations, columns = years, values = number of recors
records_per_year = (
    df.groupby(["indicativo", "year"])
    .size()
    .unstack(fill_value=0) # years become columns; 0 where no records exist
)

fig, ax = plt.subplots(figsize=(18, 10))
sns.heatmap(
    records_per_year,
    cmap="YlGn", # yellow = few recors, dark green = full coverage
    linewidths=0.3,
    linecolor="white",
    ax=ax,
    cbar_kws={"label": "Records per year"}
)

ax.set_title("Record availability by station and year")
ax.set_xlabel("Year")
ax.set_ylabel("Station")
plt.tight_layout()
plt.show()
```

## 3. Temperature Analysis

```python
STATIONS = ["4452", "4410X", "4244X", "4358X", "4468X"]

# Working subset: only the selected stations
df_sel = df[df["indicativo"].isin(STATIONS)].copy()

print(f"Records in selection: {len(df_sel)}")
print(f"\nRecords per station:")
print(df_sel["indicativo"].value_counts())
```

```python
# Flag years with fewer than 10 months of data as incomplete
months_available = (
    df_sel.groupby(["indicativo", "year"])["month"]
    .count()
)

# Keep only station-year combinations with at least 10 months
valid_year_station = months_available[months_available >= 10].index

# Filter the working dataframe to exclude incomplete years
"""
df_sel_complete = df_sel.set_index(["indicativo", "year"]).loc[
    df_sel.set_index(["indicativo", "year"]).index.isin(valid_year_station)
].reset_index()
"""
df_sel["n_months"] = df_sel.groupby(["indicativo", "year"])["month"].transform("count")
df_sel_complete = df_sel[df_sel["n_months"] >= 10].drop(columns="n_months")

removed = len(df_sel) - len(df_sel_complete)
print(f"Records removed (incomplete years): {removed}")
print(f"Records remaining: {len(df_sel_complete)}")
```

```python
# Mean of monthly average temperatures grouped by station and year
annual_temp = (
    df_sel_complete.groupby(["indicativo", "year"])["tm_mes"]
    .mean()
    .unstack("indicativo")
)

ax = annual_temp.plot(linewidth=1.5, marker="o", markersize=3)
ax.set_title("Annual mean temperature by station")
ax.set_xlabel("Year")
ax.set_ylabel("Temperature (°C)")
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Count how many months are available per station and year
months_per_year = (
    df_sel_complete.groupby(["indicativo", "year"])["month"]
    .count()
    .unstack("indicativo")
)

# Show years with fewer than 10 months for any station
incomplete = months_per_year[months_per_year.min(axis=1) < 10]
print(incomplete)
```

```python
# Average temperature profile per calendar month (all years combined)
monthly_temp = (
    df_sel_complete.groupby(["indicativo", "month"])["tm_mes"]
    .mean()
    .unstack("indicativo")
)

ax = monthly_temp.plot(linewidth=2, marker="o")
ax.set_title("Mean temperature by month (seasonal profile)")
ax.set_xlabel("Month")
ax.set_ylabel("Temperature (°C)")
ax.set_xticks(range(1, 13))
ax.set_xticklabels(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Annual mean of absolute m"aximum and minimum temperatures
extremes = df_sel_complete.groupby(["indicativo", "year"])[["ta_max", "ta_min"]].mean()

fig, axes = plt.subplots(1, 2, figsize=(16, 5), sharey=False)

for station, group in extremes.groupby("indicativo"):
    group = group.droplevel("indicativo")
    axes[0].plot(group.index, group["ta_max"], label=station, linewidth=1.5)
    axes[1].plot(group.index, group["ta_min"], label=station, linewidth=1.5)

axes[0].set_title("Mean annual absolute maximum temperature")
axes[0].set_ylabel("Temperature (°C)")
axes[0].set_xlabel("Year")
axes[0].legend(title="Station")

axes[1].set_title("Mean annual absolute minimum temperature")
axes[1].set_ylabel("Temperature (°C)")
axes[1].set_xlabel("Year")
axes[1].legend(title="Station")

plt.tight_layout()
plt.show()
```

## 4. Precipitation Analysis

```python
# Sum monthly precipitation per station and year, then pivot stations to columns
annual_precip = (
    df_sel_complete.groupby(["indicativo", "year"])["p_mes"]
    .sum()
    .unstack("indicativo")
)

ax = annual_precip.plot(linewidth=1.5, marker="o", markersize=3)
ax.set_title("Annual total precipitation by station")
ax.set_xlabel("Year")
ax.set_ylabel("Precipitation (mm)")
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Average precipitation per calendar month across all years (seasonal profile)
monthly_precip = (
    df_sel_complete.groupby(["indicativo", "month"])["p_mes"]
    .mean()
    .unstack("indicativo")
)

ax = monthly_precip.plot(linewidth=2, marker="o")
ax.set_title("Mean precipitation by month (seasonal profile)")
ax.set_xlabel("Month")
ax.set_ylabel("Precipitation (mm)")
ax.set_xticks(range(1, 13))
ax.set_xticklabels(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# A month is considered dry if total precipitation is below 10mm
dry_months = (
    (df_sel_complete["p_mes"] < 10)
    .groupby([df_sel_complete["indicativo"], df_sel_complete["year"]])
    .sum()
    .unstack("indicativo")
)

ax = dry_months.plot(linewidth=1.5, marker="o", markersize=3)
ax.set_title("Number of dry months per year (< 10 mm)")
ax.set_xlabel("Year")
ax.set_ylabel("Dry months")
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Maximum single-day precipitation recorded each year per station
annual_pmax = (
    df_sel_complete.groupby(["indicativo", "year"])["p_max"]
    .max()
    .unstack("indicativo")
)

ax = annual_pmax.plot(linewidth=1.5, marker="o", markersize=3)
ax.set_title("Annual maximum single-day precipitation")
ax.set_xlabel("Year")
ax.set_ylabel("Precipitation (mm)")
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

## 5. Wind Analysis

```python
# Average wind speed per calendar month across all years
monthly_wind = (
    df_sel_complete.groupby(["indicativo", "month"])["w_med"]
    .mean()
    .unstack("indicativo")
)

ax = monthly_wind.plot(linewidth=2, marker="o")
ax.set_title("Mean wind speed by month (seasonal profile)")
ax.set_xlabel("Month")
ax.set_ylabel("Wind speed (km/h)")
ax.set_xticks(range(1, 13))
ax.set_xticklabels(["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"])
ax.legend(title="Station", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
# Check distribution of w_racha_dir values for both stations
for station in ["4452", "4244X"]:
    vals = df_sel_complete.loc[
        df_sel_complete["indicativo"] == station, "w_racha_dir"
    ].dropna()
    print(f"\nStation {station} — value counts (top 10):")
    print(vals.value_counts().head(10))
    print(f"Unique values: {vals.nunique()}")
```

```python
fig, axes = plt.subplots(1, 2, subplot_kw={"projection": "polar"}, figsize=(12, 5))

selected = ["4452", "4244X"]

for ax, station in zip(axes, selected):
    # Get wind direction data for this station, drop NaN
    directions = df_sel_complete.loc[
        df_sel_complete["indicativo"] == station, "w_racha_dir"
    ].dropna()

    # Divide 360 degrees into 16 sectors of 22 degrees each approximately
    bins = range(0, 361, 22)
    counts, bin_edges = pd.cut(directions, bins=bins, right=False).value_counts(sort=False).align(
        pd.Series(0, index=pd.cut(directions, bins=bins, right=False).cat.categories)
    )

    # Convert sector centers to radians for the polar plot
    angles = [ ( (b.left + b.right) / 2 ) * (3.14159 / 180) for b in counts.index]

    ax.bar(angles, counts.values, width=0.35, alpha=0.7, color="steelblue")
    ax.set_title(f"Station {station}", pad=15)
    ax.set_theta_zero_location("N")   # North at the top
    ax.set_theta_direction(-1)        # Clockwise, like a compass

plt.suptitle("Wind gust direction distribution", y=1.02)
plt.tight_layout()
plt.show()
```

## 7. Conclusions

### Temperature
- The five stations show a homogeneous thermal regime consistent with the 
  Mediterranean-continental climate of Badajoz: mild winters (~8-10°C) and 
  hot summers (~26-27°C).
- Station 4358X is consistently the warmest and 4468X the coolest, suggesting 
  differences in altitude or local exposure.
- A mild warming trend is visible from 2010 onwards, particularly in 4358X 
  and 4452, though the inter-annual variability is high.
- Mean monthly temperature is a strong predictor of hot days (nt_30, r=0.92) 
  and negatively correlated with relative humidity (r=-0.86).

### Precipitation
- Precipitation is highly variable both inter-annually (200-800mm/year) and 
  spatially across the province.
- The seasonal profile is clearly Mediterranean: dry summers (Jul-Aug ~3-5mm) 
  and rainfall concentrated in autumn-winter-spring.
- Station 4468X shows an exceptionally high October peak (~103mm mean), 
  suggesting greater exposure to cold drop (DANA) events.
- Station 4358X is the driest overall; 4244X and 4468X show the most extreme 
  single-day precipitation events.
- Maximum daily precipitation and total monthly precipitation are strongly 
  correlated (r=0.83): wetter months also produce more intense episodes.

### Wind
- Stations 4452 and 4468X are the windiest (~10-12 km/h mean), while 4358X 
  is the calmest (~6-7 km/h).
- Wind is stronger in spring (March-July) and weaker in autumn, consistent 
  with the synoptic patterns of the Iberian Peninsula.
- Dominant wind gust direction is west-northwest to south-southwest for both 
  analyzed stations, reflecting the influence of Atlantic weather systems.
- Wind speed shows no significant correlation with temperature or 
  precipitation (r~0.07), confirming it is driven by independent synoptic factors.

### Data quality notes
- Records before 2008 are sparser and have fewer published variables. 
  Analysis restricted to years with at least 10 months of data.
- Fields added by AEMET after ~2015 (inso, n_llu, n_tor, nv_*) have >70% 
  missingness and were excluded from this analysis.
- w_racha_dir is encoded by AEMET on a 1-36 scale (each unit = 10°) and was 
  converted to compass degrees during the cleaning phase.
